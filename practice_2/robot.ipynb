```python
def get_sensor_data_type(value):

    if "°C" in value:
        return "Temperature"

    elif "visual fit" in value:
        return "Visual_Fit"

    elif "defect-free" in value:
        return "Defect_Free_Percentage"

    elif "obstacle detected" in value:
        return "Obstacle"

    elif "no obstacle" in value:
        return "Obstacle"

    elif "clear path" in value:
        return "Path_Clear"

    else:
        return "Unknown"
```

chi squared vtest

```python
cat_features = [f for f in df_cleaned_copy.columns if df_cleaned_copy[f].dtype == 'int64']
```

```python
cat_features
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05


chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned_copy[col], df_cleaned_copy['Defect_Detected'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
selected_features = chi2_df.loc[
    chi2_df["Decision"] == "Reject Null (Keep Feature)"
].index.tolist()

dropped_features = chi2_df.loc[
    chi2_df["Decision"] == "Accept Null (Drop Feature)"
].index.tolist()

print("KEEP:", selected_features)
print("DROP:", dropped_features)
```

```python
DROP=['Task_Type_Painting', 'Sensor_Type_LIDAR + Camera', 'Task_Type_Assembly', 'Sensor_Data_Type_Path_Clear', 'Sensor_Type_Camera', 'Sensor_Data_Type_Defect_Free_Percentage', 'Sensor_Data_Type_Temperature', 'Task_Type_Inspection', 'Obstacle_Detected', 'Environmental_Status', 'Task_Type_Welding', 'Sensor_Data_Type_Visual_Fit', 'Sensor_Data_Type_Obstacle', 'Sensor_Type_LIDAR']
```

```python
final_df=df_cleaned_copy.drop(columns=DROP)
```
