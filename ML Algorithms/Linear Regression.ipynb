**----------------------------------------------------------------------Linear Regression----------------------------------------------------------------**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
#Train-Test split
X_train, X_test, Y_train ,Y_test = train_test_split(X,Y,test_size=0.2, random_state=42)
```

```python
#Create model

#object of LinearRegression class
lr = LinearRegression()

#Training the model
lr.fit(X_train.values.reshape(-1, 1), Y_train)
```

```python
#Prediction
y_pred = lr.predict(X_test.values.reshape(-1, 1))
y_pred
```

```python
predicted_df = pd.DataFrame({'Actual':Y_test, "Predicted": y_pred, 'Difference': np.abs(Y_test - y_pred), 'Difference %': np.abs((Y_test-y_pred)*100)/Y_test})
```

Linear Regression algorithm made from scratch

```python

class LinNiloy:
    import numpy as np

    def __init__(self, X_train, Y_train):
        self.x = X_train
        self.y = Y_train
    def fit(self):
        m_x=np.mean(self.x)
        m_y=np.mean(self.y)
        numerator=sum((self.x[i]-m_x)*(self.y[i]-m_y) for i in range(len(self.x)))
        denominator=sum((self.x[i]-m_x)**2 for i in range(len(self.x)))
        self.m = numerator/denominator
        self.c = m_y - (self.m*m_x)
    def predict(self, x_test):
        return self.m*x_test + self.c
    def mae(self,y_test,y_pred):
        return sum((abs(y_test[i]-y_pred[i]))/len(y_test) for i in range(len(y_test))) 
    def mse(self,y_test,y_pred):
        return sum(((y_test[i]-y_pred[i])**2)/len(y_test) for i in range(len(y_test)))
    def r2_score(self,y_test,y_pred):
        numerator = sum((y_test[i]-y_pred[i])**2 for i in range(len(y_test)))
        denominator = sum((y_test[i]-np.mean(y_test))**2 for i in range(len(y_test)))
        return 1-(numerator/denominator)
    
```

```python
# mean absolute error
mae = linreg.mae(Y_test.values, y_pred)

# mean squared error
mse = linreg.mse(Y_test.values,y_pred)

#root mean squared error
rmse = np.sqrt(mse)

# R squared score 
r2 = linreg.r2_score(Y_test.values,y_pred)

# adjusted R squared score
n = len(Y_test)
p = 1  # number of features
ad_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
```

```python
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R Squared Score (R2):", r2)
print("Adjusted R Squared Score (Adj R2):", ad_r2)
```
