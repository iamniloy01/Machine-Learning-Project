## 4. Model 1: Linear Regression

Linear Regression models the relationship between the input features and the target variable as a linear function.

For multiple features, the prediction is calculated as:

$$
\hat{y} = b + w_1x_1 + w_2x_2 + \cdots + w_nx_n
$$

where:

- $\hat{y}$ is the predicted target.
- $b$ is the intercept (bias).
- $w_i$ is the coefficient associated with feature $i$.
- $x_i$ is the value of feature $i$.
- $n$ is the total number of features.

The scikit-learn `LinearRegression` model is trained on the training data and used to generate predictions for the test set.

The intercept and coefficients are examined to understand the fitted model.

### 4.1 Model Evaluation

Two metrics are used to evaluate the regression model.

**Mean Squared Error (MSE)**

MSE measures the average squared difference between actual and predicted values.

$$
\mathrm{MSE} = \frac{1}{m}\sum_{i=1}^{m}(y_i - \hat{y}_i)^2
$$

A lower MSE generally indicates smaller prediction errors.

**Coefficient of Determination ($(R^2)$)**

The $(R^2)$ score measures how much of the variation in the target variable is explained by the model.

$$
R^2 = 1 - \frac{\sum_{i=1}^{m}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{m}(y_i - \bar{y})^2}
$$

A score closer to 1 generally indicates a better fit. However, model performance should be judged using appropriate metrics and validation, not $(R^2)$ alone.

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
print("MSE:",mean_squared_error(y_test,y_pred))
print("R2:",r2_score(y_test,y_pred))
```

## 5. Feature Scaling

Gradient-based optimization algorithms can be sensitive to differences in feature magnitudes. Standardization transforms each feature to have approximately zero mean and unit standard deviation.

The transformation is:

$$
z=\frac{x-\mu}{\sigma}
$$

where \($mu$) is the mean and \($sigma$) is the standard deviation calculated from the training data.

The scaler is fitted **only on the training set** and then applied to both the training and testing sets. This prevents information from the test set from leaking into the training process.

Feature scaling is used for `SGDRegressor` and the custom gradient descent model.

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_trained_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 6. Model 2: SGDRegressor

`SGDRegressor` is a scikit-learn implementation of linear regression trained using Stochastic Gradient Descent (SGD).

Unlike standard Linear Regression, which computes a least-squares solution directly, SGD updates the model parameters iteratively using gradients.

The learning rate controls the size of each parameter update. In this implementation, `learning_rate="constant"` keeps the learning rate fixed at the value specified by `eta0`.

After training, predictions are generated for the scaled test data. The model's intercept, coefficients, MSE, and $(R^2)$ score are examined.

```python
from sklearn.linear_model import SGDRegressor
model = SGDRegressor(learning_rate='constant',eta0=0.01,max_iter=1000,random_state=42)
model.fit(X_trained_scaled,y_train)
```

```python
y_pred_gd = model.predict(X_test_scaled)
```

## 7. Model 3: Gradient Descent from Scratch

In this section, multiple linear regression is implemented using NumPy without relying on a prebuilt regression estimator.

### Prediction equation

$$
\hat{y} = Xw + b
$$

Here, \(X\) represents the feature matrix, \(w\) represents the vector of weights, and \(b\) represents the bias.

### Cost function

The algorithm minimizes the following cost function:

$$
J(w,b)=\frac{1}{2m}\sum_{i=1}^{m}(\hat{y}_i-y_i)^2
$$

### Gradients

The gradients of the cost function with respect to the weights and bias are:

$$
dw=\frac{1}{m}X^T(\hat{y}-y)
$$

$$
db=\frac{1}{m}\sum_{i=1}^{m}(\hat{y}_i-y_i)
$$

### Parameter updates

The weights and bias are updated iteratively:

$$
w:=w-\alpha dw
$$

$$
b:=b-\alpha db
$$

where \($alpha$) is the learning rate.

This implementation uses **Batch Gradient Descent**, because each update uses the full training dataset. The number of epochs and learning rate control the training process.

Finally, the trained model predicts target values for the test set, and its learned weights and bias are inspected.

```python
import numpy as np
class Gradient_descent:
    def __init__(self,epoch,learning_rate):
        self.epoch = epoch
        self.learning_rate = learning_rate
        self.weight=None
        self.bias=None
    def fit(self,X_train,y_train):
        self.weight=np.zeros(X_train.shape[1])
        self.bias = 0
        for i in range(self.epoch):
            y_pred = np.dot(X_train,self.weight )+ self.bias
            error=y_pred-y_train
            
            dw=np.dot(X_train.T,error)/len(X_train)
            db=np.mean(error)

            self.weight-=self.learning_rate*dw
            self.bias-=self.learning_rate*db
    def predict(self,X_test):
        return np.dot(X_test,self.weight)+self.bias
    def co_efficient(self):
        return self.weight
    def intercept(self):
        return self.bias
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

```python
GD= Gradient_descent(1000,0.01)
```

```python
GD.fit(X_train_sc,y_train)
```

```python
y_pred_GD=GD.predict(X_test_sc)
```
